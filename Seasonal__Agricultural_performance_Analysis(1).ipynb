## **10. Outlier Analysis**

### Outlier Detection

```python
# OUTLIER DETECTION USING IQR

def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    return len(outliers), lower_bound, upper_bound


important_numeric_columns = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Soil_Moisture_pct',
    'Fertilizer_kg_ha',
    'Pesticide_Litre_ha',
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Profit_INR',
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3',
    'Disease_Pest_Risk_pct'
]

outlier_results = []

for column in important_numeric_columns:
    count, lower, upper = detect_outliers_iqr(
        yield_df if column == 'Yield_Tonnes_Ha' else df,
        column
    )

    outlier_results.append([
        column,
        count,
        lower,
        upper
    ])

outlier_table = pd.DataFrame(
    outlier_results,
    columns=[
        'Variable',
        'Outlier_Count',
        'Lower_Bound',
        'Upper_Bound'
    ]
)

outlier_table
```

## **15. Economic Analysis**

### Creating Useful Derived Variables (Feature Engineering)

```python
# Profit margin
df['Profit_Margin_pct'] = np.where(
    df['Revenue_INR'] != 0,
    (df['Profit_INR'] / df['Revenue_INR']) * 100,
    np.nan
)

# Cost per hectare
df['Cost_per_Hectare_INR'] = np.where(
    df['Farm_Area_Hectares'] != 0,
    df['Total_Cost_INR'] / df['Farm_Area_Hectares'],
    np.nan
)

# Revenue per hectare
df['Revenue_per_Hectare_INR'] = np.where(
    df['Farm_Area_Hectares'] != 0,
    df['Revenue_INR'] / df['Farm_Area_Hectares'],
    np.nan
)

# Profit per hectare
df['Profit_per_Hectare_INR'] = np.where(
    df['Farm_Area_Hectares'] != 0,
    df['Profit_INR'] / df['Farm_Area_Hectares'],
    np.nan
)

print("Derived variables created successfully.")
```

## **21. Statistical Test**

Hypothesis

H0: Mean yield is equal across all seasons.

H1: At least one season has a different mean yield.

```python

# ONE-WAY ANOVA: YIELD ACROSS SEASONS
kharif_yield = yield_df[
    yield_df['Season'] == 'Kharif'
]['Yield_Tonnes_Ha']

rabi_yield = yield_df[
    yield_df['Season'] == 'Rabi'
]['Yield_Tonnes_Ha']

zaid_yield = yield_df[
    yield_df['Season'] == 'Zaid'
]['Yield_Tonnes_Ha']

anova_result = stats.f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("ANOVA F-statistic:", anova_result.statistic)
print("ANOVA p-value:", anova_result.pvalue)
```

```python
# Interpret ANOVA

alpha = 0.05

if anova_result.pvalue < alpha:
    print(
        "Result: Reject the null hypothesis."
    )
    print(
        "There is statistically significant evidence that "
        "mean yield differs across seasons."
    )
else:
    print(
        "Result: Fail to reject the null hypothesis."
    )
    print(
        "There is insufficient statistical evidence that "
        "mean yield differs across seasons."
    )
```

### **Kruskal-Wallis Test**

```python
kruskal_result = stats.kruskal(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("Kruskal-Wallis statistic:", kruskal_result.statistic)
print("Kruskal-Wallis p-value:", kruskal_result.pvalue)
```

```python
#  Interpretation
if kruskal_result.pvalue < 0.05:
    print(
        "There is statistically significant evidence "
        "of differences in yield distributions across seasons."
    )
else:
    print(
        "There is insufficient statistical evidence "
        "of differences in yield distributions across seasons."
    )
```

### **Seasonal Economic Statistical Test**

```python
kharif_profit = df[
    df['Season'] == 'Kharif'
]['Profit_INR'].dropna()

rabi_profit = df[
    df['Season'] == 'Rabi'
]['Profit_INR'].dropna()

zaid_profit = df[
    df['Season'] == 'Zaid'
]['Profit_INR'].dropna()

profit_anova = stats.f_oneway(
    kharif_profit,
    rabi_profit,
    zaid_profit
)

print("Profit ANOVA F-statistic:", profit_anova.statistic)
print("Profit ANOVA p-value:", profit_anova.pvalue)
```
